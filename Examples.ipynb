```python
from DRaD import update_data, stereo_a, filefinder, utils
import numpy as np

data = stereo_a.load(month = ['2022-09', '2022-10', '2022-11', '2022-12'])

data.loc[data['V_N'] < -1000,'V_N'] = np.nan
data.loc[data['V_T'] < -1000,'V_T'] = np.nan

for col in data.columns:
    data.loc[data[col] < -1000, col] = np.nan


utils.plot_timeseries(data)
```

```python
from DRaD import stereo_a, utils
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

start_date = '2014-03'
end_date = '2024-12'

# Generate a date range with monthly frequency
months = pd.date_range(start=start_date, end=end_date, freq='M').strftime('%Y-%m').tolist()

for month in months:
    stereo_a_data = stereo_a.load(month= month)#, reduce_cadence='3h', interpolate=True)
    for col in stereo_a_data.columns:
        stereo_a_data.loc[abs(stereo_a_data[col]) > 1e20, col] = np.nan
        stereo_a_data = stereo_a_data[~stereo_a_data.index.duplicated(keep='first')]
    stereo_a_data.to_parquet('reduced_data\stereo_a\stereo_a_data'+month+'.parquet')
    utils.plot_timeseries(stereo_a_data)
    #plt.show()
    plt.savefig(f'stereo_data/monthly_plots/stereo_a_{month}.png')
    plt.close()
```

```python
from DRaD import maven, utils, filefinder
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

start_date = '2014-03'
end_date = '2024-12'

# Generate a date range with monthly frequency
months = pd.date_range(start=start_date, end=end_date, freq='M').strftime('%Y-%m').tolist()


for month in months:
    file = filefinder.find_parquet_files(root_dir = 'reduced_data\maven', months = month)
    if len(file)>0:
        #print(file)
        df = pd.read_parquet(file)
        df.to_parquet('reduced_data\maven\maven_data'+month+'.parquet')
```
